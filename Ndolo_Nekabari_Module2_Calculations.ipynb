## Question 1: Design Envelope

### Question

What is a design envelope, and why is the pressure-temperature (P-T) envelope particularly important for subsea pipeline integrity?

### Answer

A design envelope is the multidimensional safe operating space within which an asset is designed to operate. It defines acceptable operating limits for parameters such as pressure, temperature, flow rate, and environmental loads.

The pressure-temperature (P-T) envelope is particularly important for subsea pipeline integrity because pressure affects the mechanical stresses acting on the pipeline, while temperature influences material behaviour and flow assurance conditions.

Operating outside the specified limits may increase the risk of structural failure, equipment damage, or flow assurance problems, such as hydrate formation.

The design envelope helps engineers establish safe operating conditions and identify when an asset is approaching or exceeding its operating limits.


```python
# Question 1: Design Envelope

# Illustrative design envelope limits
# Replace these example values with limits from the course materials.
pressure_min_psi = 500
pressure_max_psi = 4467

temperature_min_c = 4
temperature_max_c = 82


def check_design_envelope(pressure_psi, temperature_c):
    """
    Check whether pressure and temperature are within
    the specified design envelope.
    """

    pressure_within_limits = (
        pressure_min_psi <= pressure_psi <= pressure_max_psi
    )

    temperature_within_limits = (
        temperature_min_c <= temperature_c <= temperature_max_c
    )

    if pressure_within_limits and temperature_within_limits:
        return "Operating conditions are within the design envelope."

    return "Operating conditions are outside the design envelope."


# Example operating conditions
operating_pressure_psi = 3500
operating_temperature_c = 15

# Display the result
result = check_design_envelope(
    operating_pressure_psi,
    operating_temperature_c
)

print(f"Operating pressure: {operating_pressure_psi} psi")
print(f"Operating temperature: {operating_temperature_c} °C")
print(result)
```

## Question 2: Pipeline Wall Thickness Using Barlow's Formula

### Question

Using Barlow's formula, calculate the required pipeline wall thickness for an internal pressure of 4,467 psi, an outside diameter of 10.75 inches, and a material yield strength of 65,000 psi.

### Answer

Barlow's formula for the simplified calculation without a design factor is:

$$
t = \frac{PD}{2S}
$$

Where:

* \(t\) = wall thickness in inches
* \(P\) = internal pressure in psi
* \(D\) = outside diameter in inches
* \(S\) = material yield strength in psi

Given:

* Pressure = 4,467 psi
* Outside diameter = 10.75 inches
* Yield strength = 65,000 psi

Substituting into the formula:

$$
t = \frac{4467 \times 10.75}{2 \times 65000}
$$

$$
t = 0.3694 \text{ inches}
$$

Converting to millimetres:

$$
t = 0.3694 \times 25.4
$$

$$
\boxed{t \approx 9.38 \text{ mm}}
$$

Therefore, the calculated theoretical wall thickness is approximately **9.38 mm**.

This calculation excludes design factors, corrosion allowance, manufacturing tolerances, and other code-specific requirements. It is not a final pipeline design specification.


```python
# Question 2: Calculate pipeline wall thickness using Barlow's formula

pressure_psi = 4467
outside_diameter_in = 10.75
yield_strength_psi = 65000

wall_thickness_in = (
    pressure_psi * outside_diameter_in
) / (2 * yield_strength_psi)

wall_thickness_mm = wall_thickness_in * 25.4

print(f"Wall thickness: {wall_thickness_in:.4f} inches")
print(f"Wall thickness: {wall_thickness_mm:.2f} mm")
```

## Question 3: Hydrate Risk and Subcooling

### Answer

Hydrate formation is a flow assurance risk in subsea pipelines. Under suitable pressure, temperature, and fluid-composition conditions, gas hydrates can form and potentially restrict or block pipeline flow.

The hydrate equilibrium temperature is calculated using the simplified equation provided in the course:

$$
T_{\text{hydrate}}=4.0+18.0\log_{10}\left(\frac{P}{200}\right)
$$

Where:

* \(T_{\text{hydrate}}\) = hydrate equilibrium temperature in °C
* \(P\) = pressure in psi

**Given:**

* Pressure = 3,500 psi
* Actual fluid temperature = 15°C

### Step 1: Calculate the hydrate equilibrium temperature

Substituting the values:

$$
T_{\text{hydrate}}=4.0+18.0\log_{10}\left(\frac{3500}{200}\right)
$$

$$
T_{\text{hydrate}}=4.0+18.0\log_{10}(17.5)
$$

$$
\boxed{T_{\text{hydrate}}\approx26.37^\circ C}
$$

### Step 2: Calculate subcooling

Subcooling is the difference between the hydrate equilibrium temperature and the actual fluid temperature.

$$
\text{Subcooling}=T_{\text{hydrate}}-T_{\text{actual}}
$$

$$
\text{Subcooling}=26.37-15
$$

$$
\boxed{\text{Subcooling}\approx11.37^\circ C}
$$

### Step 3: Determine the hydrate risk classification

Using the risk classification provided in the course:

| Subcooling    | Risk classification |
| ------------- | ------------------- |
| ≤ 0°C         | No risk             |
| > 0°C to 3°C  | Low risk            |
| > 3°C to 7°C  | Medium risk         |
| > 7°C to 12°C | High risk           |
| > 12°C        | Critical risk       |

Since the calculated subcooling is approximately 11.37°C, the hydrate risk classification is:

$$
\boxed{\text{High Risk}}
$$

**Conclusion:** The hydrate equilibrium temperature is approximately **26.37°C**, the subcooling is approximately **11.37°C**, and the resulting hydrate risk classification is **High risk** under the simplified classification system provided in the course.

*Note: The equation is a simplified instructional model. Actual hydrate prediction requires appropriate fluid-composition data and validated thermodynamic calculations.*


```python
import math

# Given operating conditions
pressure_psi = 3500
actual_temperature_c = 15

# Calculate hydrate equilibrium temperature
hydrate_temperature_c = (
    4.0 + 18.0 * math.log10(pressure_psi / 200)
)

# Calculate subcooling
subcooling_c = hydrate_temperature_c - actual_temperature_c

# Classify hydrate risk
if subcooling_c <= 0:
    risk = "No risk"
elif subcooling_c <= 3:
    risk = "Low risk"
elif subcooling_c <= 7:
    risk = "Medium risk"
elif subcooling_c <= 12:
    risk = "High risk"
else:
    risk = "Critical risk"

# Display results
print(f"Hydrate equilibrium temperature: {hydrate_temperature_c:.2f} °C")
print(f"Subcooling: {subcooling_c:.2f} °C")
print(f"Hydrate risk classification: {risk}")
```

### Question

Identify four internal corrosion mechanisms that may affect subsea pipelines. Explain their driving factors and identify relevant monitoring parameters or sensors.

### Answer

Internal corrosion can reduce pipeline wall thickness, weaken structural integrity, and eventually contribute to leakage or failure.

Four relevant mechanisms are:

**1. CO₂ corrosion**

Carbon dioxide dissolves in water and can create an acidic environment that promotes corrosion. Its severity depends on factors such as CO₂ partial pressure, temperature, water chemistry, and flow conditions.

Monitoring parameters include CO₂ partial pressure, temperature, water cut, pH, and corrosion rate.

**2. H₂S-related corrosion**

Hydrogen sulfide can contribute to corrosion and, under susceptible conditions, hydrogen damage such as sulfide stress cracking. Risk depends on H₂S exposure, material properties, stress, and environmental conditions.

Monitoring parameters include H₂S concentration or partial pressure, temperature, water chemistry, and material condition.

**3. Erosion-corrosion**

Flowing fluids and entrained particles can remove protective surface films and accelerate material loss. High flow velocities and sand production may increase the risk.

Monitoring parameters include flow velocity, sand rate, pressure, wall thickness, and corrosion rate.

**4. Microbiologically influenced corrosion (MIC)**

Microorganisms may contribute to localized corrosion through biological activity and changes in the local chemical environment.

Monitoring methods include microbial sampling, water-chemistry analysis, pH measurement, temperature monitoring, and corrosion coupons or probes.

Monitoring these parameters helps engineers assess corrosion conditions and identify developing integrity problems.

## Question 5: Corrosion Rate and Remaining Life

### Question

A pipeline's wall thickness has decreased from 12.7 mm to 10.8 mm over 10 years. If the minimum allowable wall thickness is 8.5 mm, calculate the average corrosion rate and estimate the remaining life.

### Answer

Given:

* Initial wall thickness = 12.7 mm
* Current wall thickness = 10.8 mm
* Inspection period = 10 years
* Minimum allowable wall thickness = 8.5 mm

**Step 1: Calculate wall-thickness loss.**

$$
\text{Wall loss} = 12.7 - 10.8 = 1.9\text{ mm}
$$

**Step 2: Calculate average corrosion rate.**

$$
\text{Corrosion rate} = \frac{\text{Wall loss}}{\text{Inspection period}}
$$

$$
\text{Corrosion rate} = \frac{1.9}{10}
$$

$$
\boxed{0.19\text{ mm/year}}
$$

**Step 3: Calculate remaining wall-thickness allowance.**

$$
\text{Remaining allowance} = 10.8 - 8.5
$$

$$
\boxed{2.3\text{ mm}}
$$

**Step 4: Estimate remaining life.**

Assuming a constant corrosion rate:

$$
\text{Remaining life} = \frac{\text{Remaining allowance}}{\text{Corrosion rate}}
$$

$$
\text{Remaining life} = \frac{2.3}{0.19}
$$

$$
\boxed{\text{Remaining life} \approx 12.1\text{ years}}
$$

Therefore, the estimated remaining life is approximately **12.1 years**, assuming the corrosion rate remains constant.

This estimate is not a guaranteed service life. Actual corrosion rates can change, so inspection planning should consider corrosion trends, measurement uncertainty, failure consequences, and an appropriate integrity assessment.


```python
# Question 5: Calculate corrosion rate and estimated remaining life

initial_thickness_mm = 12.7
current_thickness_mm = 10.8
minimum_thickness_mm = 8.5
inspection_period_years = 10

wall_loss_mm = initial_thickness_mm - current_thickness_mm

corrosion_rate_mm_per_year = (
    wall_loss_mm / inspection_period_years
)

remaining_allowance_mm = (
    current_thickness_mm - minimum_thickness_mm
)

remaining_life_years = (
    remaining_allowance_mm / corrosion_rate_mm_per_year
)

print(f"Wall-thickness loss: {wall_loss_mm:.2f} mm")
print(f"Average corrosion rate: {corrosion_rate_mm_per_year:.2f} mm/year")
print(f"Remaining allowance: {remaining_allowance_mm:.2f} mm")
print(f"Estimated remaining life: {remaining_life_years:.2f} years")
```

## Question 7: Fatigue Life Using an S-N Curve

### Question

Using the D seawater cathodic-protection S-N curve, calculate the fatigue life at a stress range of 60 MPa.

### Answer

An S-N curve describes the relationship between stress range and the number of loading cycles a material can withstand before fatigue failure.

According to the course, the parameters for the `D_seawater_cp` curve are:

$$
\log_{10}(a) = 11.764
$$

$$
m = 3.0
$$

The fatigue-life relationship is:

$$
\log_{10}(N) = \log_{10}(a) - m\log_{10}(\Delta\sigma)
$$

Where:

* \(N\) = fatigue life in cycles
* \(\log_{10}(a)\) = S-N curve constant
* \(m\) = S-N curve slope parameter
* \(\Delta\sigma\) = stress range in MPa

Given:

$$
\Delta\sigma = 60\text{ MPa}
$$

Substituting:

$$
\log_{10}(N) = 11.764 - 3.0\log_{10}(60)
$$

$$
\log_{10}(N) \approx 6.4295
$$

Therefore:

$$
N = 10^{6.4295}
$$

$$
\boxed{N \approx 2.69 \times 10^6\text{ cycles}}
$$

The estimated fatigue life is approximately **2.69 million cycles** under the selected S-N relationship.


```python
# Question 7: Calculate fatigue life using the S-N curve

import math

sn_curves = {
    "B1_air": {"log_a": 15.117, "m": 4.0},
    "D_air": {"log_a": 12.164, "m": 3.0},
    "D_seawater_cp": {"log_a": 11.764, "m": 3.0},
    "E_air": {"log_a": 12.010, "m": 3.0},
    "F_air": {"log_a": 11.855, "m": 3.0}
}

curve_name = "D_seawater_cp"
stress_range_mpa = 60

params = sn_curves[curve_name]

log_n = (
    params["log_a"]
    - params["m"] * math.log10(stress_range_mpa)
)

fatigue_life_cycles = 10 ** log_n

print(f"S-N curve: {curve_name}")
print(f"Stress range: {stress_range_mpa} MPa")
print(f"Estimated fatigue life: {fatigue_life_cycles:.0f} cycles")
```

## Question 8: Cumulative Fatigue Damage Using Miner's Rule

### Question

A component experiences fatigue-damage contributions of 0.15, 0.08, and 0.03 under three different loading conditions. Calculate the total cumulative fatigue damage. If this damage accumulated uniformly over five years, estimate the total fatigue life and remaining life.

### Answer

Miner's rule estimates cumulative fatigue damage by adding the damage contributions from different loading conditions:

$$
D = \sum_{i=1}^{n}\frac{n_i}{N_i}
$$

Where:

* \(D\) = cumulative fatigue-damage ratio
* \(n_i\) = number of cycles experienced at a particular stress range
* \(N_i\) = estimated fatigue life at that stress range

**Step 1: Calculate cumulative fatigue damage.**

$$
D = 0.15 + 0.08 + 0.03
$$

$$
\boxed{D = 0.26}
$$

The accumulated damage is 0.26, equivalent to 26% of the theoretical fatigue-damage limit under the simplified Miner’s rule approach.

**Step 2: Estimate total fatigue life.**

Assuming the damage accumulated uniformly over five years:

$$
\text{Total fatigue life} = \frac{\text{Elapsed time}}{\text{Accumulated damage}}
$$

$$
\text{Total fatigue life} = \frac{5}{0.26}
$$

$$
\boxed{\text{Total fatigue life} \approx 19.23\text{ years}}
$$

**Step 3: Estimate remaining fatigue life.**

$$
\text{Remaining life} = 19.23 - 5
$$

$$
\boxed{\text{Remaining life} \approx 14.23\text{ years}}
$$

The estimated remaining life is approximately **14.23 years**, assuming future damage accumulates at the same average rate.


```python
# Question 8: Calculate cumulative fatigue damage using Miner's rule

damage_contributions = [0.15, 0.08, 0.03]
elapsed_years = 5

total_damage = sum(damage_contributions)

estimated_total_life_years = elapsed_years / total_damage

remaining_life_years = (
    estimated_total_life_years - elapsed_years
)

print(f"Total cumulative damage: {total_damage:.2f}")
print(f"Accumulated damage: {total_damage * 100:.1f}%")
print(f"Estimated total fatigue life: {estimated_total_life_years:.2f} years")
print(f"Estimated remaining life: {remaining_life_years:.2f} years")
```

## Question 10: Engineering Data Dictionary

### Question

Identify the essential fields that should be included in an engineering data dictionary and explain why each field is important.

### Answer

A data dictionary is a structured reference that describes the variables in a dataset. It helps engineers and data scientists understand the meaning, units, data types, and acceptable values of the data used in analysis and machine-learning models.

The essential fields are:

1. **Variable name:** Identifies the variable in the dataset, for example, `pressure_psi`.
2. **Data type:** Defines the type of value stored, such as `float64` for numerical measurements.
3. **Unit:** Specifies the measurement unit, such as psi, °C, or barrels/day.
4. **Physical meaning:** Explains what the variable represents, such as pipeline operating pressure.
5. **Valid operating range:** Defines the expected minimum and maximum values for the variable.
6. **Data source:** Identifies where the measurement or value originates, such as a pressure transmitter or flow meter.

### Example

| Variable name   | Data type | Unit        | Physical meaning                   | Valid range |
| --------------- | --------- | ----------- | ---------------------------------- | ----------- |
| `pressure_psi`  | `float64` | psi         | Operating pressure                 | 0–5,000     |
| `temperature_c` | `float64` | °C          | Operating temperature              | -5–100      |
| `flow_rate_bpd` | `float64` | barrels/day | Fluid flow rate                    | 0–40,000    |
| `water_cut_pct` | `float64` | %           | Water percentage in produced fluid | 0–100       |

A well-maintained data dictionary improves consistency, supports data validation, and reduces errors caused by inconsistent units, misunderstood variables, or invalid values.

For predictive AI projects, it also provides a foundation for feature engineering, model development, and future maintenance.
